Q7. Sum of weights

```python
asia_cars = df[df['origin'] == 'Asia']
asia_selected = asia_cars[['vehicle_weight', 'model_year']].head(7)
asia_selected
```

```python
X = asia_selected.to_numpy()

XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
w
```

```python
w = w.sum()
w
```
